```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df[['Time', 'Amount']] = scaler.fit_transform(df[['Time', 'Amount']])
```

```python
df.head()
```

```python
X = df.drop('Class', axis =1)
y = df['Class']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42)
```

```python
from imblearn.over_sampling import SMOTE

sm = SMOTE(random_state=42)
X_train,y_train = sm.fit_resample(X_train,y_train)
```

```python
plt.subplot(1,2,2)
sns.countplot(x=y_train)
plt.title("After SMOTE")
plt.xlabel("Class")
plt.ylabel("Count")

plt.tight_layout()
plt.show()
```

```python
df_smote = X_train.copy()
df_smote["Class"] = y_train

plt.figure(figsize=(6,5))
sns.countplot(x="Class", data=df_smote)
plt.title("Class Distribution After SMOTE")
plt.xlabel("Class")
plt.ylabel("Count")
```

```python
df_smote['Class'].value_counts()
```

```python
from sklearn.linear_model import LogisticRegression

modellr = LogisticRegression(random_state=42)
modellr.fit(X_train, y_train)
```

```python
y_pred_lr = modellr.predict(X_test)
```

```python
y_pred_lr
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score

modellr_accuracy = accuracy_score(y_test, y_pred_lr)
print("Accuracy :",modellr_accuracy)
print("\nClassification Report")
print(classification_report(y_test, y_pred_lr))

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred_lr))

print("\nROC AUC Score")
print(roc_auc_score(y_test, y_pred_lr))
```

```python
from xgboost import XGBClassifier

modelxgb = XGBClassifier(
    random_state=42,
    eval_metric='logloss'
)

modelxgb.fit(X_train, y_train)
```

```python
y_pred_xgb = modelxgb.predict(X_test)
```

```python
y_pred_xgb
```

```python
modelxgb_accuracy = accuracy_score(y_test, y_pred_xgb)
print("Accuracy :",modelxgb_accuracy)
print("Classification Report",classification_report(y_test, y_pred_xgb))
print("Confusion Matrix",confusion_matrix(y_test, y_pred_xgb))
print("ROC AUC :", roc_auc_score(y_test, y_pred_xgb))
```

```python
from sklearn.ensemble import RandomForestClassifier

modelrf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
modelrf.fit(X_train, y_train)
```

```python
y_pred_rf = modelrf.predict(X_test)
```

```python
y_pred_rf
```

```python
modelrf_accuracy = accuracy_score(y_test, y_pred_rf)
print("Accuracy :",modelrf_accuracy)
print("Classification Report",classification_report(y_test, y_pred_rf))
print("Confusion Matrix",confusion_matrix(y_test, y_pred_rf))
print("ROC AUC :", roc_auc_score(y_test, y_pred_rf))
```

```python
models = ["Logistic Regression", "Random Forest", "XGBoost"]

accuracy = [
    modellr_accuracy,
    modelrf_accuracy,
    modelxgb_accuracy
]

plt.figure(figsize=(8, 5))

ax = sns.barplot(
    x=models,
    y=accuracy
)

plt.title("Model Accuracy Comparison")
plt.xlabel("Models")
plt.ylabel("Accuracy")
for i, value in enumerate(accuracy):
    plt.text(
        i,
        value + 0.005,
        f"{value * 100:.2f}%",
        ha="center"
    )
plt.show()
```

```python
import joblib

joblib.dump(modellr, "logistic_regression_model.pkl")
joblib.dump(modelrf, "random_forest_model.pkl")
joblib.dump(modelxgb, "xgboost_model.pkl")
joblib.dump(scaler, "scaler.pkl")
```
